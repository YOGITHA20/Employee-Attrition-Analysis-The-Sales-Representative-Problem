```python
df = df.drop(columns=["EmployeeCount", "StandardHours", "Over18"])
```

```python
print(df["MonthlyIncome"].describe())
```

```python
df["Attrition"] = df["Attrition"].map({"Yes": 1, "No": 0})  # convert to 1/0 for SQL-friendly math

employees = df[["EmployeeNumber","Age","Gender","MaritalStatus","Department",
                "JobRole","DistanceFromHome","OverTime","Attrition"]]
compensation = df[["EmployeeNumber","MonthlyIncome","PercentSalaryHike","StockOptionLevel"]]
performance = df[["EmployeeNumber","PerformanceRating","JobSatisfaction",
                   "WorkLifeBalance","YearsSinceLastPromotion"]]

employees.to_csv("employees.csv", index=False)
compensation.to_csv("compensation.csv", index=False)
performance.to_csv("performance.csv", index=False)
```
